```python
import pandas as pd
import joblib

def preprocess_and_predict(raw_data, model, scaler, expected_columns):
    """
    Takes raw customer data, cleans it, formats it perfectly, and returns a prediction.
    """
    # 1. Convert raw input (like a dictionary) into a Pandas DataFrame
    if isinstance(raw_data, dict):
        df = pd.DataFrame([raw_data])
    else:
        df = raw_data.copy()
        
    # 2. Apply Custom Data Cleaning (Match training phase logic)
    if 'EDUCATION' in df.columns:
        df['EDUCATION'] = df['EDUCATION'].replace([0, 5, 6], 4)
    if 'MARRIAGE' in df.columns:
        df['MARRIAGE'] = df['MARRIAGE'].replace(0, 3)
        
    # 3. One-Hot Encode categorical variables
    encoded_df = pd.get_dummies(df)
    
    # 4. Align with expected columns (adds missing ones as 0, drops extra ones)
    aligned_df = encoded_df.reindex(columns=expected_columns, fill_value=0)
    
    # 5. Scale the data
    scaled_array = scaler.transform(aligned_df)
    
    # 6. Make the prediction
    prediction = model.predict(scaled_array)
    
    return prediction[0]
```
